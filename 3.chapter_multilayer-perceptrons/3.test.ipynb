## 1.数据预处理

```python
transform = transforms.ToTensor()
# 这个就是一个转换规则，意思转换成Tensor这个数据类型，因为读进来的是原始图片，要把他转换成tensor的数据类型
```

## 2. Download Training Set

```python
train_data = datasets.FashionMNIST(
    root="./data",
    train=True,
    download=True,
    transform=transform
)
```

```python
test_data = datasets.FashionMNIST(
    root="./data",
    train=False,
    download=True,
    transform=transform
)
```

```python
image, lable = train_data[1]
print(image)
print(lable)
# 所以这个dataset这个对象其实就是一个存了数据本身和lable的一个东西
```

```python
# 创建dataloader
batch_size = 256

train_iter = DataLoader(
    train_data,
    batch_size=batch_size,
    shuffle=True
)

test_iter = DataLoader(
    test_data,
    batch_size=batch_size,
    shuffle=False
)
# 此时的train_iter变成了一个DataLoader对象
```

```python
# 如果要查看里面的数据
X, y = next(iter(train_iter))
X
y
# 这样取出来的就是第一个第一个iter
```

```python
# model
net = nn.Sequential(
    nn.Flatten(), # 将一个28*28的单个对象展平成[1, 784]
    nn.Linear(784, 256), # 中间层，将784变成256个节点
    nn.ReLU(), # 添加一个非线性层
    nn.Linear(256, 10) # 输出层，讲中间层的256输出成10个分类结果
)
```

```python
# Loss function
loss_fn = nn.CrossEntropyLoss()
```

```python
# optimizer
optimizer = torch.optim.SGD(
    net.parameters(),
    lr = 0.1
)
```

```python
net.parameters()
# 其实也就会获取w1, w2, b1, b2
```

```python
# 训练函数

def train(net, train_iter, loss_fn, optimizer, epochs):
    for epoch in range(epochs):

        net.train() # 启动train模式

        for X, y in train_iter:
            y_hat = net(X) # 前向传播，根据输入给出预测值
            loss = loss_fn(y_hat, y)
            optimizer.zero_grad()
            loss.backward() # 反向传播
            optimizer.step() # 更新参数
```

```python
# 测试函数

def test(net, test_iter):

    net.eval()

    correct  = 0
    total = 0

    with torch.no_grad(): # 意思不用计算梯度，节约资源

        for X, y in test_iter:
            y_hat = net(X)
            pred = y_hat.argmax(dim = 1)
            correct += (pred==y).sum().item() # item()是将tensor转换为可以累加的python数字
            total += y.size(0)

    return correct/total
```

```python
# 开始训练
train(net, train_iter, loss_fn, optimizer, epochs=10)

acc = test(net, test_iter)

print(acc)
```
