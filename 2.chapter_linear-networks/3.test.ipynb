```python
import numpy as np
import torch
from torch.utils import data
from d2l import torch as d2l
```

```python
true_w = torch.tensor([2, 0.1])
true_b = 1
features, lables = d2l.synthetic_data(true_w, true_b, 19)
features
lables
```

```python
data_arrays = (features, lables)
dataset = data.TensorDataset(*data_arrays)
dataset[0]
data_iter = data.DataLoader(dataset=dataset, batch_size=5, shuffle=True)
```

```python
from torch import nn
net = nn.Sequential(
    nn.Linear(2, 1) #两个输入一个输出
)
```

```python
net[0].weight.data.normal_(0, 0.01)
net[0].bias.data.fill_(0)

loss = nn.MSELoss()
```

```python
net.parameters()
```

```python
trainer = torch.optim.SGD(net.parameters(), lr = 0.03)
```

```python
num_epochs = 3
for epoch in range(num_epochs):
    for X, y in data_iter:

        l = loss(net(X), y) # 计算loss
        trainer.zero_grad() # 清空梯度
        l.backward() # 仅仅计算梯度
        trainer.step() # step()函数是执行预定步骤

    l = loss(net(features), lables)
    print(f"epoch {epoch+1}, loss{l:f}")
```
